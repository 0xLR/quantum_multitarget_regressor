```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (10,5)
```

```python
def plot_data(df, x, y):
    fig, ax = plt.subplots()
    for key, grp in df.groupby(['optimizer']):
        ax = grp.plot(ax=ax, kind='line', x=x, y=y, label=key)
    
    plt.legend(loc='best')    
    plt.show()
```

```python
df = pd.read_csv('results-final.csv')
df
```

## 100 Samples by 80/20 Split

```python
# 100 Samples by 80/20 split
df_100_samples_by_80_20_split = df[(df['n_samples']==100) & (df['test_size'] == 20)]
plot_data(df_100_samples_by_80_20_split, 'iterations', 'mae')
```

```python
# 100 Samples by 70/30 split
df_100_samples_by_70_30_split = df[(df['n_samples']==100) & (df['test_size'] == 30)]
plot_data(df_100_samples_by_70_30_split, 'iterations', 'mae')
```
