```python
from sklearn.datasets import make_regression
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
# 100 Samples
scaler = MinMaxScaler()
X, Y = make_regression(n_samples=100, n_features=3, n_targets=2, random_state=42)
X = scaler.fit_transform(X)
Y = scaler.fit_transform(Y)
samples_100 = pd.DataFrame(np.hstack((X, Y)), columns = [u'x\u2081',u'x\u2082',u'x\u2083',u'y\u2081',u'y\u2082'])
samples_100.to_html('training-100-samples.html')
```
